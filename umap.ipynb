```python
## full form : Uniform Manifold Approximation and Projection.
#use for 1. dimension reducation and visuliaze the dataset
```

```python
# pip install umap-learn scikit-learn matplotlib

import umap
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler

# Load data
digits = load_digits()
X = digits.data
y = digits.target

# Standardize
X = StandardScaler().fit_transform(X)

# UMAP: 64D -> 2D
reducer = umap.UMAP(
    n_neighbors=15,
    min_dist=0.1,
    n_components=2,
    metric="euclidean",
    random_state=42
)

X_umap = reducer.fit_transform(X)

# Plot
plt.figure(figsize=(10, 7))
plt.scatter(
    X_umap[:, 0],
    X_umap[:, 1],
    c=y,
    cmap="tab10",
    s=10
)

plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.title("UMAP Visualization of Digits Dataset")
plt.colorbar(label="Digit")
plt.show()
```
